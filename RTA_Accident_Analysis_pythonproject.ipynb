## Feature Engineering

The Time column is converted into an hour-based numerical feature so that accident patterns across different times of the day can be analyzed by the machine learning models.

```python
data['Hour'] = pd.to_datetime(
    data['Time'],
    format='%H:%M:%S',
    errors='coerce'
).dt.hour
```

```python
data[['Time', 'Hour']].head()
```

```python
data['Hour'] = pd.to_datetime(
    data['Time'],
    errors='coerce'
).dt.hour
```

```python
data['Hour'].isnull().sum()
```

```python
def time_period(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

data['Time_period'] = data['Hour'].apply(time_period)
```

##ML data set

```python
ml_data = data.copy()
```

```python
columns_to_drop = [
    'Time',
    'Casualty_class',
    'Sex_of_casualty',
    'Age_band_of_casualty',
    'Casualty_severity',
    'Work_of_casuality',
    'Fitness_of_casuality',
    'Pedestrian_movement'
]

ml_data = ml_data.drop(columns=columns_to_drop)
```

## Separating Features and Target

```python
X = ml_data.drop('Accident_severity', axis=1)

y = ml_data['Accident_severity']
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
y.value_counts()
```

##Split into training and testing data

## Train-Test Split

The dataset is divided into training and testing sets. The training data is used to train the machine learning models, while the testing data is used to evaluate their performance on unseen data.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
#Check the split
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

##Encode categorical variables

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
categorical_features = X.select_dtypes(
    include=['object']
).columns.tolist()
```

```python
numerical_features = X.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()
```

```python
print("Categorical columns:")
print(categorical_features)

print("\nNumerical columns:")
print(numerical_features)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=False
            ),
            categorical_features
        ),
        (
            'numerical',
            'passthrough',
            numerical_features
        )
    ]
)
```

## Machine Learning Model 1 — Logistic Regression

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
```

```python
lr_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(
        max_iter=1000,
        class_weight='balanced'
    ))
])
```

```python
class_weight='balanced'
```

```python
lr_model.fit(X_train, y_train)
```

```python
y_pred_lr = lr_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, f1_score

print("Logistic Regression Accuracy:",
      accuracy_score(y_test, y_pred_lr))

print("\nLogistic Regression Macro F1:",
      f1_score(y_test, y_pred_lr, average='macro'))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_lr))
```

## Machine Learning Model 2 — Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', DecisionTreeClassifier(
        random_state=42,
        class_weight='balanced',
        max_depth=10
    ))
])
```

```python
dt_model.fit(X_train, y_train)
```

```python
y_pred_dt = dt_model.predict(X_test)
```

```python
print("Decision Tree Accuracy:",
      accuracy_score(y_test, y_pred_dt))

print("\nDecision Tree Macro F1:",
      f1_score(y_test, y_pred_dt, average='macro'))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt))
```

## Machine Learning Model 3 — Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight='balanced',
        n_jobs=-1
    ))
])
```

```python
rf_model.fit(X_train, y_train)
```

```python
y_pred_rf = rf_model.predict(X_test)
```

```python
print("Random Forest Accuracy:",
      accuracy_score(y_test, y_pred_rf))

print("\nRandom Forest Macro F1:",
      f1_score(y_test, y_pred_rf, average='macro'))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

##Compare all 3 models

```python
results = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Decision Tree',
        'Random Forest'
    ],
    'Accuracy': [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_rf)
    ],
    'Macro F1': [
        f1_score(y_test, y_pred_lr, average='macro'),
        f1_score(y_test, y_pred_dt, average='macro'),
        f1_score(y_test, y_pred_rf, average='macro')
    ]
})

results
```

```python
results.set_index('Model')[['Accuracy', 'Macro F1']].plot(
    kind='bar',
    figsize=(10,6)
)

plt.title('Machine Learning Model Comparison')
plt.ylabel('Score')
plt.ylim(0,1)

plt.xticks(rotation=0)

plt.show()
```

```python
#finding the best model according to Macro F1:
results.sort_values(
    by='Macro F1',
    ascending=False
)
```

```python
best_predictions = y_pred_dt
best_model_name = 'Decision Tree'

print("Best Model:", best_model_name)
print("Accuracy:", accuracy_score(y_test, best_predictions))
print("Macro F1:", f1_score(y_test, best_predictions, average='macro'))
```

```python
print(classification_report(y_test, best_predictions))
```

## Confusion Matrix

The confusion matrix shows the actual accident severity compared with the severity predicted by the Decision Tree model.

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    best_predictions,
    labels=['Slight Injury', 'Serious Injury', 'Fatal injury']
)

plt.figure(figsize=(8,6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    xticklabels=['Slight Injury', 'Serious Injury', 'Fatal injury'],
    yticklabels=['Slight Injury', 'Serious Injury', 'Fatal injury']
)

plt.title('Confusion Matrix - Decision Tree')
plt.xlabel('Predicted Severity')
plt.ylabel('Actual Severity')

plt.show()
```

```python
print("Final Model:", best_model_name)
print(f"Accuracy: {accuracy_score(y_test, best_predictions):.4f}")
print(f"Macro F1 Score: {f1_score(y_test, best_predictions, average='macro'):.4f}")
```
